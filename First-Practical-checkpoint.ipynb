```python
import pandas as pd

data = {
    "Student": ["Ali", "Sara", "Ahmed"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

students = ["Ali", "Sara", "Ahmed"]
marks = [85, 90, 78]

plt.bar(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```
